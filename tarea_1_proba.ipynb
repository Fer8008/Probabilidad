##Ejercicio 1

`Al lanzar dos dados`, calcula la probabilidad de que la suma de sus caras sea exactamentem 5.

```python
from itertools import product # modulo para producto cartesiano

# definimos que es una list creando tuplas
espacio_m = list(product(range(1,7), repeat= 2))
# funcion product generar todas las combinaciones posibles de los dados(2 porque son 2 dados)
print(f"Espacio Muestral: {espacio_m}")

# filtra suma la caras x[0] + x[1] de x == 5
casos_f = [x for x in espacio_m if x[0] + x[1] == 5]
print(f"Casos favorables: {casos_f}")

# cuenta casos_f, espacio_m y hace la division
print("Probabilidad:", len(casos_f)/len(espacio_m)) # cuenta casos_f, espacio_m y hace la division
```

##Ejercicio 2
En una ``contraseña de 3 dígitos``, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.


```python
from itertools import product # modulo para producto cartesiano

# creamos list y repetimos 3 veces porque son 3 numeros de 0 al 9
espacio_m = list(product(range(10), repeat=3))

# conjuntos=set(x) elimina duplicados, por lo que su longitud debe ser 3
# ninguno repetirse si se repite seria una tupla y la elimina
casos_f = [x for x in espacio_m if len(set(x)) == 3]

# al ser el resultado una lista muy larga utilizamos len para que solo nos de el total
print(f"Total de resultados posibles: {len(espacio_m)}")
print(f"Total de casos favorables: {len(casos_f)}")
print(f"Probabilidad: {len(casos_f) / len(espacio_m)}")
```

##Ejercicio 3
Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

```python
from itertools import permutations

# creamos espacio muestral con todas las permutaciones posibles de 8 canciones (del 1 al 8)
# la canción favorita seria la 1 y la que menos gusta es la 8.
espacio_m = list(permutations(range(1, 9)))

# filtramos los casos favorables
# x[0] == 1 (la favorita queda primera) y x[-1] == 8 (la menos favorita queda última)
casos_f = [x for x in espacio_m if x[0] == 1 and x[-1] == 8]

# mostramos los resultados totales de posibles, favorables y la probabilidad
print(f"Total de resultados posibles: {len(espacio_m)}")
print(f"Total de casos favorables: {len(casos_f)}")
print(f"Probabilidad: {len(casos_f) / len(espacio_m)}")
```

#Ejercicio 4
Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?

```python
from itertools import product

# el espacio muestral de 3 dados de 6 caras (del 1 al 6)
# se repite 3 veces porque son 3 dados
espacio_m = list(product(range(1, 7), repeat=3))

# casos favorables
# el ciclo for recorre la lista espacio_m y busca que las suma de sean 9 o 10
casos_9 = [x for x in espacio_m if sum(x) == 9]
casos_10 = [x for x in espacio_m if sum(x) == 10]

print(f"Total de resultados posibles: {len(espacio_m)}")
print(f"Probabilidad de la suma 9: {len(casos_9) / len(espacio_m)}")
print(f"Probabilidad de la suma 10: {len(casos_10) / len(espacio_m)}")
print(f"Casos favorables con suma 9 es de {len(casos_9)}: {casos_9}")
print(f"Casos favorables con suma 10 es de {len(casos_10)}: {casos_10}")
# utilizamos abs(valor absoluto) para que el resultado siempre sea un numero positvo
print(f"La suma ganadora por {abs(len(casos_9) - len(casos_10))} fue la del numero 10")
```

#Ejercicio 5
Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.

```python
from itertools import permutations

# lista de 4 amigos
personas = (0, 1, 2, 3)

# generamos todas las formas posibles de repartir los regalos permutaciones dentro de una lista
espacio_m = list(permutations(personas))

# la funcion all revisa la condicion que nadien tenga su propio regalo
# en la tupla resultante debe de ser diferente i  de  i (x[i] != i) en el rango de la lista
casos_f = [x for x in espacio_m if all(x[i] != i for i in range(len(personas)))]

print(f"Total de resultados posibles: {len(espacio_m)}")
print(f"Total de casos favorables: {len(casos_f)}")
print(f"Probabilidad: {len(casos_f)} / {len(espacio_m)} = {len(casos_f) / len(espacio_m)}")
```
