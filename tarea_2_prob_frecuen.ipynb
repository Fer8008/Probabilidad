##Ejercicio 1
Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

```python
import random
import matplotlib.pyplot as plt

def lanzar_dado():
    return random.randint(1, 6)

simulaciones = 10000
conteo_suma_5 = 0
frecuencias = []

for i in range(1, simulaciones + 1):
    suma = lanzar_dado() + lanzar_dado()
    if suma == 5:
        conteo_suma_5 += 1

    frecuencia = conteo_suma_5 / i
    frecuencias.append(frecuencia)

print(f"Numero de simulaciones: {simulaciones}")
print(f"Probabilidad empírica: {frecuencia}")
print(f"Probabilidad clásica teórica: {4/36}")

# grafica
plt.figure(figsize=(12, 8))
plt.plot(frecuencias, label="Frecuencia relativa")
plt.axhline(4/36, color='black', linestyle="--", label="Probabilidad clásica (4/36 = 1/9)")
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
plt.title("Probabildad empirica vs Probabilidad Clásica [Suma 5]")
plt.legend()
plt.show()
```

##Ejercicio 2
Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

```python
import random
import matplotlib.pyplot as plt

# generamos un pines con el número de dígitos indicado
def generar_pin(digitos):
    return [random.randint(0, 9) for _ in range(digitos)]

simulaciones = 10000
conteo_sin_repetir = 0
frecuencias = []

for i in range(1, simulaciones + 1):
    pin = generar_pin(4)

    # si la longitud del conjunto es igual a 4, ninguno se repite
    if len(set(pin)) == 4:
        conteo_sin_repetir += 1

    frecuencia = conteo_sin_repetir / i
    frecuencias.append(frecuencia)

print(f"Número de simulaciones: {simulaciones}")
print(f"Frecuencia de PIN sin dígitos repetidos: {frecuencia}")

# Gráfica de convergencia
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Frecuencia relativa")
plt.axhline(5040/10000, color= "black", linestyle="--", label="Probabilidad clásica = 5040/10000 (0.504)")
plt.xlabel("Número de simulaciones")
plt.ylabel("Frecuencia relativa")
plt.title("PIN de 4 dígitos sin repetir")
plt.legend()
plt.show()
```

##Ejercicio 3
Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

```python
import random
import matplotlib.pyplot as plt

def lanzar_dado():
    return random.randint(1, 6)

n = 20000
conteo_9 = 0
conteo_10 = 0

frecuencias_9 = []
frecuencias_10 = []

for i in range(1, n + 1):
    suma = lanzar_dado() + lanzar_dado() + lanzar_dado()

    if suma == 9:
        conteo_9 += 1
    if suma == 10:
        conteo_10 += 1

    frecuencias_9.append(conteo_9 / i)
    frecuencias_10.append(conteo_10 / i)

print("Número de lanzamientos:", n)
print("Frecuencia empírica de suma 9:", conteo_9 / n)
print("Frecuencia empírica de suma 10:", conteo_10 / n)

plt.figure(figsize=(10, 5))
plt.plot(frecuencias_9, label="Frecuencia relativa (Suma 9)")
plt.plot(frecuencias_10, label="Frecuencia relativa (Suma 10)", color='orange')
plt.axhline(25/216, linestyle="--", color='blue', label="Teórica Suma 9 (25/216)")
plt.axhline(27/216, linestyle=":", color='orange', label="Teórica Suma 10 (27/216)")
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion Suma 9 vs Suma 10 en tres dados")
plt.legend()
plt.show()
```

##Ejercicio 4
Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
import random
import matplotlib.pyplot as plt

# definimos una función que simula extraer una bola de la urna
def sacar_bola():
    urna = ["roja", "roja", "roja", "azul", "azul"]
    return random.choice(urna)

# numero total de extracciones en el experimento
repetir = 2000

#variable para contar cuántas veces obtenemos una bola azul
#empezando en 0
conteo_azules = 0

# lista donde guardaremos la frecuencia relativa después de cada extracción
frecuencias = []

# repetimos el experimento 2000 veces
for i in range(1, repetir + 1):
    resultado = sacar_bola()

    # verificamos si el resultado es azul
    if resultado == "azul":
        conteo_azules += 1 # si es azul aumenta en 1

 # calculamos la frecuencia relativa (azules observadas / total de extracciones)
    frecuencia = conteo_azules / i
    frecuencias.append(frecuencia)

print(f"Número de extracciones: {repetir}")
print(f"Probabilidad empírica de obtener una bola azul: {frecuencia}")


# graficamos
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Frecuencia relativa de bola azul")
plt.axhline(2/5, color= "black", linestyle="--", label="Probabilidad teórica = 2/5 (0.4)")
plt.xlabel("Número de extracciones")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad empírica (Bola azul)")
plt.legend()
plt.show()
```

##Ejercicio 5
Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
import random
import matplotlib.pyplot as plt

# función que simula el lanzamiento de una moneda
def lanzar_moneda():
    return random.choice(["águila", "sol"])

# función que genera una serie de 20 lanzamientos
def simular_serie():
    return [lanzar_moneda() for _ in range(20)]

# numero total simulaciones
n = 10000

# contador de series donde SI apareció la racha de 5 aguilas
conteo_exitos = 0

# lista para guardar la frecuencia relativa en cada paso
frecuencias = []

# repetimos el experimento n veces
for i in range(1, n + 1):
    serie = simular_serie()

    # variables para detectar la racha dentro de la serie
    racha_actual = 0
    tiene_racha_5 = False

    for tiro in serie:
        if tiro == "águila":
            racha_actual += 1
            if racha_actual >= 5:
                tiene_racha_5 = True
                break # encontramos la racha, no necesitamos seguir buscando en esta serie
        else:
            racha_actual = 0  # si sale sol, se reinicia la racha

    if tiene_racha_5:
        conteo_exitos += 1

    # calculamos la frecuencia relativa acumulada
    frecuencia = conteo_exitos / i
    frecuencias.append(frecuencia)

print("Número de series simuladas:", n)
print("Probabilidad estimada de 5 águilas seguidas en 20 volados:", frecuencia)

# gráfica de los resultados
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Frecuencia relativa de racha de 5 águilas")
plt.xlabel("Número de series (de 20 volados c/u)")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad empírica: Racha de 5 águilas seguidas en 20 lanzamientos")
plt.legend()
plt.show()
```

##Ejercicio 6 Bonus
Los clientes llegan a una tienda de forma aleatoria entre 1 y 5 minutos. El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
import random
import matplotlib.pyplot as plt

# número total de clientes a simular
n = 500

# variables iniciales para el control del tiempo en la tienda
tiempo_llegada_acumulado = 0
tiempo_libre_cajero = 0
conteo_espera_mayor_10 = 0

# lista para guardar la frecuencia relativa después de cada cliente
frecuencias = []

# repetimos el experimento para cada uno de los n clientes
for i in range(1, n + 1):
    # tiempo aleatorio entre la llegada del cliente anterior que es entre 1 y 5 min.
    tiempo_entre_llegadas = random.uniform(1, 5)
    tiempo_llegada_acumulado += tiempo_entre_llegadas

    # tiempo aleatorio que tarda el cajero en atender a este cliente es entre 2 y 4 min.
    tiempo_atencion = random.uniform(2, 4)

    # evaluamos si el cliente tiene que esperar en la fila
    if tiempo_llegada_acumulado >= tiempo_libre_cajero:
        # el cajero está libre y el cliente no espera nada
        tiempo_espera = 0
        tiempo_libre_cajero = tiempo_llegada_acumulado + tiempo_atencion
    else:
        # El cajero está ocupado y el cliente calcula cuánto debe esperar
        tiempo_espera = tiempo_libre_cajero - tiempo_llegada_acumulado
        tiempo_libre_cajero += tiempo_atencion

    # verificamos si el tiempo de espera fue mayor a 10 minutos
    if tiempo_espera > 10:
        conteo_espera_mayor_10 += 1

    # calculamos la frecuencia relativa acumulada
    frecuencia = conteo_espera_mayor_10 / i
    frecuencias.append(frecuencia)

print("Número total de clientes simulados:", n)
print("Probabilidad empírica de esperar más de 10 minutos:", frecuencia)

# Gráfica de los resultados
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Frecuencia relativa (Espera > 10 min)")
plt.xlabel("Número de clientes")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad de espera mayor a 10 minutos en fila")
plt.legend()
plt.show()
```
