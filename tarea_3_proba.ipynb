#*Probabilidad condicional*

##Ejercicio 1
Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

```python
espacio = [1, 2, 3, 4, 5, 6]
B = [x for x in espacio if x % 2 == 0]
A = [x for x in espacio if x == 2]
interseccion = set(A) & set(B)
print(f"La probabilidad de que salga un 2 dado que salio un es de= {len(interseccion) / len(B):.4f}")
```

##Ejercicio 2
Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
from itertools import product

S = list(product(range(1, 7), repeat=3))
distintos = [t for t in S if len(set(t)) == 3]
suma_9 = [t for t in distintos if sum(t) == 9]

print(f"La probabilidad de que sea 9 es de:", len(suma_9),"/", len(distintos))
```

#*Teorema de Bayes*

##Ejercicio 1
El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta».

 ¿cuál es la probabilidad de que sea spam?

```python
spam = 0.20
no_spam = 0.80
o_spam = 0.80
o_no_spam = 0.10

# probabilidad total de que tenga la palabra oferta
p_o = (o_spam * spam) + (o_no_spam * no_spam)

# aplicamos el Teorema de Bayes
oferta_spam = (o_spam * spam) / p_o
print(f"La probabilidad de que sea spam un correo con la palabra oferta es de= {oferta_spam:.3f}")
```

##Ejercicio 2
En una tienda online el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

```python
chrome= 0.40
compras_c= 0.03
safari = 0.35
compras_sa = 0.04
otros = 0.25
compras_o = 0.02

# probabilidad total de compra
p_compra_total = (compras_c * chrome) + (compras_sa * safari) + (compras_o * otros)

# aplicamos el Teorema de Bayes
compra_chrom = (compras_c * chrome) / p_compra_total
compra_safari = (compras_sa * safari) / p_compra_total
compra_otro = (compras_o * otros) / p_compra_total

print(f"La probabilidad de que un cliente compre por Chrome es de= {compra_chrom:.3f}")
print(f"La probabilidad de que un cliente compre por Safari es de= {compra_safari:.3f}")
print(f"La probabilidad de que un cliente compre por otros medios es de= {compra_otro:.3f}")
```
